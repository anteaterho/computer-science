```python
a = [1,2,3,4,5,6]
b = []
for i in a:
  b.append(i * 100)

print(b)
```

```python
a = [[1,2,3],
     [4,5,6],
     [7,8,9]]
b = [[],[],[]]

for i in a:
  for j in i:
    b.append(j * 5)

print(b)
```

```python
a = [[1,2,3],
     [4,5,6],
     [7,8,9]]
b = []

for i in a:
  tmp=[]
  for j in i:
    tmp.append(j*5)
  b.append(tmp)

print(b)

# 리스트로 접근하면 행렬을 게산할 수 없다.
```
